```python
df_1 = pd.DataFrame({'ID' : [1,2,3,4,5,],
                     'class' : [9,10,11,12,13]})

df_1
```

```python
df_2 = pd.DataFrame({'ID' : [6,7,8,9,10],
                     'name' : [14,15,16,17,18]})
df_2
```

```python
merge =pd.merge(df_1,df_2,on = 'ID')
merge
```

```python
merge =pd.merge(df_1,df_2,on = 'ID' , how = 'inner')
merge
```

```python
merge =pd.merge(df_1,df_2,on = 'ID' , how = 'left')
merge
```

```python
merge =pd.merge(df_1,df_2,on = 'ID' , how = 'right')
merge
```

```python
merge =pd.merge(df_1,df_2,on = 'ID' , how = 'outer')
merge
```

```python
merge =pd.merge(df_1,df_2,on = 'ID' , how = 'outer', indicator = True)
merge
```

```python
merge =pd.merge(df_1,df_2,on = 'ID' , how = 'inner', indicator = True)
merge
```

```python
merge =pd.merge(df_1,df_2,on = 'ID' , how = 'left', indicator = True)
merge
```

```python
merge =pd.merge(df_1,df_2,on = 'ID' , how = 'right', indicator = True)
merge
```

```python
merge =pd.merge(df_1,df_2,on = 'ID' , how = 'outer', indicator = True)
merge
```

```python
data = {

        'age' : [23,34,56,np.nan,45,22,np.nan],
        'name' : ["meet", "raj", "yash", "bhima", "meera", "mansi", "vivek"],
        'city' : ['raj', 'jun', 'ahe', 'jam', 'por', 'kes', 'sur']
}
df = pd.DataFrame(data)
print("orignal data \n")
df
```
