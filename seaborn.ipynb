```python
tips = sns.load_dataset('tips')
tips
```

```python
sns.lineplot(x='total_bill', y='tip', data=tips)
```

```python
sns.lineplot(x = "size" , y = "total_bill", data = tips, hue = "sex")
```

```python
sns.lineplot(x = "time" , y = "total_bill", data = tips, hue = "sex")
```

```python
sns.lineplot(x = "size" , y = "total_bill", data = tips, hue = "sex", style = "sex")
```

```python
sns.lineplot(x = "size" , y = "total_bill", data = tips, hue = "sex", style = "sex", palette= "hot")
```

```python
sns.lineplot(x = "size" , y = "total_bill", data = tips, hue = "sex", style = "sex", palette= "hot" , dashes=False)
```

```python
sns.lineplot(x = "size" , y = "total_bill", data = tips, hue = "sex", style = "sex", palette= "hot" , dashes=False , markers=['o', '<'], legend="brief")
```

```python
import matplotlib.pyplot as plt
sns.lineplot(x = "size",
             y = "total_bill",
             data = tips,
             hue = "day",
             style = "day",
             palette= "hot" ,
             dashes=False)


plt.title('Line plot', fontsize = 20)
plt.xlabel('size', fontsize = 15)
plt.ylabel('total_bill', fontsize = 15)
plt.show()
```
