## Ploting Bar chart

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib import style
```

```python
classes = ['Python', 'R', 'AI', 'ML', 'DS']
class1_students = [30,10,20,25,15]
class2_students = [4,5,20,20,10]
class3_students = [3,5,30,15,15]
```

```python
plt.bar(classes, class1_students, color = 'y')
```

```python
plt.barh(classes, class1_students)
```

```python
plt.bar(classes, class1_students, width = 0.5, align = 'edge',color = 'y')
```

```python
plt.bar(classes, class1_students, width = 0.2, align = 'edge',color = 'm', edgecolor = 'r')
```

```python
plt.bar(classes, class1_students, width = 0.2, align = 'edge',color = 'm', edgecolor = 'b', linewidth = 5, alpha = 0.5)
```

```python
plt.bar(classes, class1_students, width = 0.2, align = 'edge',color = 'y',
        edgecolor = 'r', linewidth = 0.5, alpha = 0.5, linestyle = '--')
```

```python
plt.bar(classes, class1_students, width = 0.2, align = 'edge',color = 'y',
        edgecolor = 'r', linewidth = 0.5, alpha = 0.5, linestyle = '--',
       label = 'class 1 students',  visible = True)
plt.legend()
```

```python
plt.bar(classes, class1_students, width = 0.2, align = 'edge',color = 'b',
        edgecolor = 'b', linewidth = 0.5, alpha = 0.5, linestyle = '--',
       label = 'class 1 students')
```

```python
style.use('ggplot')
plt.bar(classes, class1_students, width = 0.2, align = 'edge',color = 'k',
        edgecolor = 'r', linewidth = 0.5, alpha = 0.5, linestyle = '--',
       label = 'class 1 students')
```

```python
style.use('ggplot')
plt.figure(figsize = (16,9))
plt.bar(classes, class1_students, width = 0.2, align = 'edge',color = 'k',
        edgecolor = 'r', linewidth = 0.5, alpha = 0.9, linestyle = '--',
       label = 'class 1 students')
```

```python
style.use('ggplot')
plt.figure(figsize = (16,9))
plt.barh(classes, class1_students, align = 'edge',color = 'k',
        edgecolor = 'r', linewidth = 0.5, alpha = 0.5, linestyle = '--',
       label = 'class 1 students')
```

```python
plt.figure(figsize = (16,9))
plt.bar(classes, class1_students, align = 'edge',color = 'b',
        edgecolor = 'r', linewidth = 0.2, alpha = 0.2, linestyle = '--',
       label = 'class 1 students')


plt.title('Bar chart of MU class', fontsize = 15)
plt.xlabel('classes', fontsize = 20)
plt.ylabel('No.of students', fontsize = 20)
plt.show()
```

```python

classes = ['Python', 'R', 'AI', 'ML', 'DS']
class1_students = [30,10,20,25,15]
class2_students = [4,5,20,20,10]
class3_students = [3,5,30,15,15]


plt.figure(figsize = (16,9))

width = 0.4
plt.bar(classes, class1_students, width  ,color = 'b',
       label = 'class 1 students')

plt.bar(classes, class2_students,width, color = 'm',
         label = 'class 2 students')


plt.title('Bar chart of MU class', fontsize = 15)
plt.xlabel('classes', fontsize = 20)
plt.ylabel('No.of students', fontsize = 20)
plt.show()
```

```python
plt.figure(figsize = (16,9))

classes_index = np.arange(len(classes))

width = 0.2
plt.bar(classes_index, class1_students,width, color = 'b',
       label = 'class 1 students')

plt.bar(classes_index + width, class2_students,width, color = 'm',
         label = 'class 2 students')


plt.title('Bar chart of MU class', fontsize = 18)
plt.xlabel('classes', fontsize = 20)
plt.ylabel('No.of students', fontsize = 20)
plt.show()
```

```python
plt.figure(figsize = (16,9))

classes_index = np.arange(len(classes))

width = 0.2
plt.bar(classes_index, class1_students,width, color = 'b',
       label = 'class 1 students')

plt.bar(classes_index + width, class2_students,width, color = 'm',
         label = 'class 2 students')

plt.xticks(classes_index + width, classes, rotation

           = 20) #x label manipulation


plt.title('Bar chart of MU class', fontsize = 18)
plt.xlabel('classes', fontsize = 20)
plt.ylabel('No.of students', fontsize = 20)
plt.show()
```

```python
plt.figure(figsize = (16,9))

classes_index = np.arange(len(classes))

width = 0.2
plt.bar(classes_index, class1_students,width, color = 'b',
       label = 'class 1 students')

plt.bar(classes_index + width, class2_students,width, color = 'm',
         label = 'class 2 students')

plt.bar(classes_index + width + width, class3_students,width, color = 'y',
         label = 'class 3 students')

plt.xticks(classes_index + width, classes, rotation = 20) #x label manipulation


plt.title('Bar chart of MU class', fontsize = 18)
plt.xlabel('classes', fontsize = 20)
plt.ylabel('No.of students', fontsize = 20)
plt.show()
```
